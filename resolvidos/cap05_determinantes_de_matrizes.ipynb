## 5.1 Calculando o determinante · PDF p. 72

#### Exercício: Determinante na mão

Escreva `det2(M)` que calcula o determinante de uma matriz 2×2 pela fórmula ad − bc. Compare com `np.linalg.det` para `[[3, 8], [4, 6]]`.

```python
def det2(M):
    (a, b), (c, d) = M
    return a * d - b * c

M = [[3, 8], [4, 6]]
print(det2(M), np.linalg.det(M))
```

#### Exercício: Determinante zero

Calcule o determinante de `[[2, 4], [1, 2]]` e explique, num comentário, o que um determinante zero diz sobre a inversa e sobre o sistema linear.

```python
S = np.array([[2, 4], [1, 2]])
print(np.linalg.det(S))
# det = 0: a matriz não tem inversa e o sistema não tem solução única
```

#### Exercício: Propriedades

Com matrizes aleatórias 3×3, confira: det(A·B) = det(A)·det(B) e det(Aᵀ) = det(A). Depois veja que o determinante de uma triangular é o produto da diagonal.

```python
rng = np.random.default_rng(1)
A = rng.integers(-5, 6, (3, 3)); B = rng.integers(-5, 6, (3, 3))
d = np.linalg.det
print(np.isclose(d(A @ B), d(A) * d(B)), np.isclose(d(A.T), d(A)))
T = np.triu(A)
print(np.isclose(d(T), np.prod(np.diag(T))))
```
