#### 🧪 Exercício: Sistema 2×2

Resolva com `np.linalg.solve`: 2x + y = 7 e x − y = −1. Confira substituindo a solução nas equações.

```python
A = np.array([[2, 1], [1, -1]]); b = np.array([7, -1])
x = np.linalg.solve(A, b)
print(x)
print(A @ x, "deve ser", b)
```

#### 🧪 Exercício: Problema dos ingressos

Um cinema vendeu 200 ingressos e arrecadou R$ 5.200. Inteira custa R$ 30 e meia, R$ 15. Monte o sistema e descubra quantos de cada tipo foram vendidos.

```python
# i + m = 200 ; 30i + 15m = 5200
A = np.array([[1, 1], [30, 15]]); b = np.array([200, 5200])
inteiras, meias = np.linalg.solve(A, b)
print(f"inteiras = {inteiras:.0f}, meias = {meias:.0f}")
```

#### 🧪 Exercício: Sistema 3×3 pela inversa

Resolva x + y + z = 6, 2y + 5z = −4, 2x + 5y − z = 27 de dois jeitos: `solve` e `inv(A) @ b`. Os resultados batem?

```python
A = np.array([[1, 1, 1], [0, 2, 5], [2, 5, -1]]); b = np.array([6, -4, 27])
print(np.linalg.solve(A, b))
print(np.linalg.inv(A) @ b)
```
