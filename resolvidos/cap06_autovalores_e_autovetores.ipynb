## 6.1 Transformação linear · PDF p. 78

#### 🧪 Exercício: Transformar um quadrado

Os cantos de um quadrado são as colunas de `Q = [[0, 1, 1, 0, 0], [0, 0, 1, 1, 0]]`. Aplique a escala `[[2, 0], [0, 0.5]]` e o cisalhamento `[[1, 1], [0, 1]]` e desenhe os três contornos com `plt.plot`.

```python
Q = np.array([[0, 1, 1, 0, 0], [0, 0, 1, 1, 0]])
E = np.array([[2, 0], [0, 0.5]]); C = np.array([[1, 1], [0, 1]])
for M, rot in [(np.eye(2), "original"), (E, "escala"), (C, "cisalhamento")]:
    P = M @ Q
    plt.plot(P[0], P[1], label=rot)
plt.axis("equal"); plt.legend(); plt.show()
```

## 6.2 Autovalores e autovetores de uma matriz · PDF p. 79

#### 🧪 Exercício: Autovalores e autovetores

Calcule autovalores e autovetores de `A = [[2, 1], [1, 2]]` com `np.linalg.eig` e confira, para cada par, que A·v = λ·v.

```python
A = np.array([[2, 1], [1, 2]])
valores, vetores = np.linalg.eig(A)
for lam, v in zip(valores, vetores.T):
    print(lam, v, np.allclose(A @ v, lam * v))
```

#### 🧪 Exercício: Remontar a matriz

Com os autovetores (colunas de P) e autovalores (diagonal de D) do exercício anterior, confira que A = P·D·P⁻¹.

```python
A = np.array([[2, 1], [1, 2]])
valores, P = np.linalg.eig(A)
D = np.diag(valores)
print(np.allclose(A, P @ D @ np.linalg.inv(P)))
```
