#### Exercício: É simétrica?

Escreva `eh_simetrica(A)` que devolve `True` quando `A` é igual à sua transposta. Teste com `[[1, 7], [7, 2]]` (sim) e `[[1, 2], [3, 4]]` (não).

```python
def eh_simetrica(A):
    A = np.asarray(A)
    return A.shape[0] == A.shape[1] and np.array_equal(A, A.T)

print(eh_simetrica([[1, 7], [7, 2]]), eh_simetrica([[1, 2], [3, 4]]))
```

## 3.4 Operações entre vetores · PDF p. 48

#### Exercício: Produto escalar, norma e ângulo

Para `u = [3, 4]` e `w = [4, 3]`: calcule o produto escalar (`np.dot`), a norma de cada um (`np.linalg.norm`) e o ângulo entre eles em graus, usando cos θ = u·w / (|u||w|).

```python
u = np.array([3, 4]); w = np.array([4, 3])
prod = np.dot(u, w)
nu, nw = np.linalg.norm(u), np.linalg.norm(w)
angulo = np.degrees(np.arccos(prod / (nu * nw)))
print(prod, nu, nw, round(angulo, 2))
```

#### Exercício: Vetores ortogonais

Dois vetores são ortogonais quando o produto escalar é zero. Verifique quais pares são ortogonais: `([1, 2], [-2, 1])`, `([1, 1], [1, -2])`, `([2, 0, 1], [0, 5, 0])`.

```python
pares = [([1, 2], [-2, 1]), ([1, 1], [1, -2]), ([2, 0, 1], [0, 5, 0])]
for a, b in pares:
    print(a, b, "ortogonais" if np.dot(a, b) == 0 else "não ortogonais")
```

## 3.6 Operações entre matrizes e vetores · PDF p. 50

#### Exercício: Total da compra

A matriz `P` do exercício anterior (preços por loja) e o vetor `q = [2, 6, 1]` (quantidades). Calcule quanto a compra custa em cada loja com `q @ P` e diga qual loja é mais barata.

```python
P = np.array([[10, 12], [5, 4.5], [30, 28]])
q = np.array([2, 6, 1])
totais = q @ P
print(totais, "-> loja mais barata:", int(np.argmin(totais)) + 1)
```

#### Exercício: Girar um ponto 90°

A matriz de rotação de 90° é `R = [[0, -1], [1, 0]]`. Aplique `R` ao ponto `(2, 1)` e depois aplique 4 vezes seguidas. O que acontece?

```python
R = np.array([[0, -1], [1, 0]])
p = np.array([2, 1])
print(R @ p)
x = p
for _ in range(4):
    x = R @ x
print(x, "-> voltou ao ponto inicial")
```

```python
rng = np.random.default_rng(0)
A = rng.integers(0, 10, (3, 3)); B = rng.integers(0, 10, (3, 3))
print(np.allclose((A @ B).T, B.T @ A.T))
```

## 3.8 Inversa de uma matriz · PDF p. 57

#### Exercício: Inversa e identidade

Calcule a inversa de `A = [[4, 7], [2, 6]]` e confira que `A @ inv(A)` dá a identidade.

```python
A = np.array([[4, 7], [2, 6]])
Ai = np.linalg.inv(A)
print(Ai)
print(A @ Ai)
print(np.allclose(A @ Ai, np.eye(2)))
```

#### Exercício: Matriz sem inversa

Tente inverter `S = [[1, 2], [2, 4]]`. Use `try/except np.linalg.LinAlgError` para mostrar uma mensagem amigável. Por que ela não tem inversa? (dica: compare as linhas)

```python
S = np.array([[1, 2], [2, 4]])
try:
    np.linalg.inv(S)
except np.linalg.LinAlgError:
    print("S não tem inversa: a 2ª linha é o dobro da 1ª (matriz singular)")
```
