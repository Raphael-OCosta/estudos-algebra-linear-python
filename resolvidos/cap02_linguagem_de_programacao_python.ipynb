## 2.6 Funcionalidades da linguagem Python · PDF p. 27

#### 🧪 Exercício: Média sem NumPy

Escreva uma função `media(valores)` que recebe uma lista de números e devolve a média, usando só `for` (sem `sum` nem NumPy). Teste com `[4, 8, 15, 16, 23, 42]` (resposta: 18.0).

```python
def media(valores):
    total = 0
    for v in valores:
        total += v
    return total / len(valores)

print(media([4, 8, 15, 16, 23, 42]))
```

#### 🧪 Exercício: Lista vs. array

Crie a lista `l = [1, 2, 3]` e o array `a = np.array([1, 2, 3])`. Compare `l * 2` com `a * 2` e `l + l` com `a + a`. Escreva num comentário por que os resultados são diferentes.

```python
l = [1, 2, 3]
a = np.array([1, 2, 3])
print(l * 2, a * 2)   # lista repete; array multiplica cada elemento
print(l + l, a + a)   # lista concatena; array soma elemento a elemento
```
